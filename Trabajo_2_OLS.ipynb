## Introducción

El presente estudio analiza la relación entre la autonomía fiscal de los gobiernos locales y su nivel de gasto social en el Perú, empleando modelos supervisados de regresión lineal (OLS).
El propósito general es comprender cómo la capacidad de los municipios para generar ingresos propios, junto con la dependencia de las transferencias intergubernamentales, incide sobre el volumen de recursos destinados a políticas sociales.
Desde una perspectiva empírica, se busca evaluar el comportamiento de estas variables a través de modelos econométricos comparativos que permitan interpretar, con rigor y evidencia, la dinámica fiscal subnacional.

La autonomía fiscal constituye un pilar fundamental en la gestión descentralizada, pues refleja el grado de independencia financiera que tienen los gobiernos locales respecto del gobierno central.
Un municipio con alta autonomía depende en menor medida de las transferencias y puede tomar decisiones más flexibles sobre la asignación de su presupuesto, mientras que aquellos con autonomía reducida suelen depender de fondos externos, lo que puede limitar su capacidad de priorizar el gasto según las necesidades locales.
En este contexto, el gasto social representa un indicador relevante del esfuerzo gubernamental por mejorar las condiciones de vida de la población, especialmente en los ámbitos de educación, salud, saneamiento e infraestructura básica.

El análisis se sitúa dentro de la literatura sobre **finanzas públicas y federalismo fiscal**, la cual reconoce dos fuentes principales de financiamiento municipal.
Por un lado, las **transferencias intergubernamentales**, que incluyen el Fondo de Compensación Municipal (FONCOMUN), el canon, sobrecanon, regalías, renta de aduanas y participaciones, buscan reducir desigualdades fiscales entre territorios.
Sin embargo, una dependencia excesiva de ellas puede generar **riesgo moral**, desincentivar el esfuerzo recaudatorio y disminuir la rendición de cuentas ciudadana.
Por otro lado, los **ingresos propios** —derivados de tributos municipales como el predial, arbitrios, licencias, multas o servicios— fortalecen la autonomía, promueven eficiencia en la asignación del gasto y refuerzan la relación entre contribuyentes y servicios públicos.  
El presente trabajo examina cómo estos dos mecanismos interactúan y se reflejan en el gasto social.

## Variables clave

Las principales variables del análisis son:

- **Autonomía fiscal:** mide la proporción de ingresos propios sobre el total de ingresos del municipio. Expresa el grado de independencia financiera de cada gobierno local.
- **Proporción de transferencias:** representa el porcentaje de los ingresos municipales que proviene del gobierno central a través de FONCOMUN, canon y otras fuentes de transferencias.
- **Gasto social:** corresponde al gasto ejecutado en funciones vinculadas al bienestar (educación, salud, programas sociales, infraestructura básica). Para reducir la dispersión y estabilizar la varianza, se utiliza el **logaritmo natural del gasto social** (*ln gasto social*).

Estas variables se integran en modelos de regresión lineal con el objetivo de estimar cómo la autonomía y las transferencias inciden, conjunta o separadamente, sobre el gasto social.

## 2. Preparación y limpieza de datos

En esta etapa se realizan los procesos de depuración y preprocesamiento del conjunto de datos.
Esto incluye la selección de variables relevantes, la creación de nuevas variables derivadas (como la autonomía fiscal y el logaritmo del gasto social), y el manejo de valores faltantes o inconsistentes.




```python
# Variable dependiente: gasto social (log transformado para estabilizar varianza)
df["ln_gasto_social"] = np.log1p(df["gasto_social"]) # log(1+x) evita log(0)
# Visualización
df[["anio", "mes", "autonomia", "prop_transferencias", "gasto_social", "ln_gasto_social"]].head()
```


### Definición de las variables explicativas y del objetivo

En este bloque se definen las variables que formarán parte de los modelos de regresión.
El **objetivo (target)** del análisis es el *logaritmo natural del gasto social*, que representa el nivel de inversión social ejecutado por cada gobierno local.
Las **variables explicativas (features)** corresponden a los determinantes fiscales principales: la *autonomía fiscal* y la *proporción de transferencias* recibidas del gobierno central.

Posteriormente, los datos se dividen en dos subconjuntos:
- un conjunto de **entrenamiento (75%)**, que se utiliza para estimar los modelos y ajustar los parámetros; y
- un conjunto de **prueba (25%)**, reservado para evaluar la capacidad predictiva fuera de muestra.

Esta separación permite garantizar que las métricas de desempeño reflejen la generalización del modelo y no un simple ajuste a los datos observados.



```python
# Variables explicativas principales
TARGET = "ln_gasto_social"
FEATURES = ["autonomia", "prop_transferencias"]

X = df[FEATURES]
y = df[TARGET]

# División train-test (75%-25%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print(f"Tamaño training: {len(X_train)}")
print(f"Tamaño test: {len(X_test)}")
```


### Establecimiento del modelo base (baseline)

Antes de estimar modelos econométricos, se calcula una **referencia inicial o baseline** que predice el valor promedio del gasto social en el conjunto de entrenamiento.
Este modelo ingenuo no utiliza información de las variables explicativas; simplemente asume que todas las observaciones tendrán el mismo valor medio.

El propósito de este paso es contar con un punto de comparación:
cualquier modelo que no logre mejorar el error del baseline (medido mediante el **Error Cuadrático Medio, MSE**) no aporta valor explicativo adicional.
Por tanto, este primer cálculo permite verificar que los modelos OLS posteriores realmente ofrecen un mejor desempeño y una interpretación económica más sólida.


```python
# Baseline: predice la media del gasto social logarítmico
mean_baseline = y_train.mean()
baseline_mse = mean_squared_error(y_train, [mean_baseline]*len(y_train))

print("\n--- Baseline ---")
print(f"Predicción baseline (media): {mean_baseline:.4f}")
print(f"MSE baseline (training): {baseline_mse:.4f}")
```

El modelo baseline predice el valor promedio del gasto social.
Servirá como punto de comparación: cualquier modelo con MSE mayor sería peor que simplemente usar la media.

## 3. Modelo OLS lineal

En esta sección se estima un modelo de **regresión lineal ordinaria (OLS)** con el objetivo de analizar la relación directa entre la autonomía fiscal y el gasto social.
El modelo adopta una forma funcional lineal, lo que implica que cada incremento en la autonomía fiscal se asocia con un cambio constante en el gasto social, manteniendo controlada la proporción de transferencias.

El propósito de esta estimación inicial es establecer un punto de comparación para modelos más complejos y evaluar la magnitud e interpretación de los coeficientes.
Se reportan los resultados obtenidos a través de la librería `statsmodels`, que incluyen los valores estimados de los parámetros, su significancia estadística y el coeficiente de determinación (R²).



```python
# Modelo Simple: gasto_social ~ autonomia + prop_transferencias
X_train_sm_simple = sm.add_constant(X_train)
model_simple = sm.OLS(y_train, X_train_sm_simple).fit()
print("\n--- OLS Simple (Statsmodels) ---")
print(model_simple.summary())
```

### Interpretación de Resultados (Modelo Simple)

Los resultados del modelo de regresión lineal simple muestran una relación positiva y significativa entre la autonomía financiera y el gasto social, así como entre la proporción de transferencias y el gasto social.

**Coeficientes:**  
- El coeficiente de **autonomía (84.35)** indica que, manteniendo constante la proporción de transferencias, un incremento en la autonomía fiscal se asocia con un aumento en el gasto social.  
- El coeficiente de **prop_transferencias (17.73)** también es positivo, lo que sugiere que los municipios que reciben más transferencias tienden, en promedio, a gastar más en educación y salud.

**P-values:**  
Ambas variables presentan valores p < 0.05, por lo que son **estadísticamente significativas**. Esto respalda la hipótesis de que la autonomía y las transferencias influyen de manera relevante en el nivel de gasto social.

**R-cuadrado ajustado (0.961):**  
El modelo explica aproximadamente el **96% de la variabilidad del gasto social**, lo que indica un muy buen ajuste del modelo a los datos.

**Diagnósticos:**  
No se observan problemas graves en los diagnósticos básicos, aunque los valores extremos en las pruebas de normalidad (Omnibus y Jarque-Bera) sugieren que los residuos podrían no seguir una distribución perfectamente normal. Esto es común en datos financieros agregados, pero no invalida la significancia del modelo.  
En general, los resultados son consistentes con la teoría: **mayor autonomía y mayores transferencias se asocian con un mayor gasto social**.


# --- Validación Cruzada ---

```python
# Validación cruzada (k=5)
lr_simple = LinearRegression()
k = 5
mse_cv_simple = -cross_val_score(lr_simple, X_train, y_train, cv=k, scoring='neg_mean_squared_error')

print("\n--- Validación Cruzada (OLS Simple) ---")
print(f"MSE por fold (simple): {mse_cv_simple}")
print(f"MSE promedio (simple): {np.mean(mse_cv_simple):.4f} (+/- {np.std(mse_cv_simple):.4f})")
```

### Interpretación de Resultados de Validación Cruzada (Modelo Simple)

El error cuadrático medio (MSE) promedio del modelo simple es **0.2141**, lo que representa el error de predicción esperado al estimar el gasto social en nuevos datos.  
Este valor sirve como referencia para evaluar si los modelos más complejos logran una mejora significativa en el rendimiento predictivo.

La **desviación estándar de 0.1076** indica una variabilidad moderada en el desempeño del modelo entre los distintos pliegues (folds) de la validación cruzada. Esto sugiere que el modelo tiene un comportamiento **relativamente estable**, aunque con cierta sensibilidad a la composición de los datos de entrenamiento.

En general, el modelo simple logra una buena capacidad predictiva, lo que refuerza que tanto la autonomía como la proporción de transferencias son variables relevantes para explicar el gasto social municipal.

## 4. Modelo OLS con forma funcional más compleja

En esta etapa se amplía la especificación del modelo OLS para capturar posibles variaciones en la intensidad de la relación entre autonomía y gasto social.
Esto se logra mediante la incorporación de un término adicional derivado de la autonomía fiscal, lo que permite explorar una forma funcional más rica dentro del marco lineal clásico.

El objetivo es analizar si una especificación más flexible mejora la capacidad explicativa del modelo, comparando los resultados con los del modelo simple en términos de ajuste y coherencia económica.
Esta comparación aporta evidencia sobre la estabilidad y robustez de la relación entre las variables fiscales.



### ¿Por qué usamos la variable “autonomía²”?

Se incluyó la variable **autonomía²** para capturar posibles **efectos no lineales** entre la autonomía financiera y el gasto social.  
La relación entre ambas puede no ser estrictamente lineal: al inicio, una mayor autonomía fiscal puede incentivar una mejor asignación de recursos y aumentar el gasto social; sin embargo, a partir de cierto punto, el efecto podría disminuir o incluso revertirse.  

Por ello, el término cuadrático permite representar una **curva en forma de ∩ (rendimientos decrecientes)**, donde el impacto positivo de la autonomía sobre el gasto social tiende a estabilizarse o reducirse en niveles altos de autonomía.


```python
#Creamos una nueva variable cuadrática de autonomía
df["autonomia2"] = df["autonomia"]**2

# Creamos los conjuntos de entrenamiento y prueba para el modelo complejo
X_train_complex = X_train.copy()
X_test_complex = X_test.copy()

X_train_complex["autonomia2"] = X_train["autonomia"]**2
X_test_complex["autonomia2"] = X_test["autonomia"]**2

# Añadimos constante e instalamos modelo con statsmodels
X_train_sm_complex = sm.add_constant(X_train_complex)
model_complex = sm.OLS(y_train, X_train_sm_complex).fit()

print("\n--- OLS Complejo (Statsmodels) ---")
print(model_complex.summary())
```

### Interpretación de Resultados (Modelo Complejo)

**Inferencia:**  
El modelo OLS Complejo, que incorpora un término cuadrático de la autonomía fiscal, presenta un **R² ajustado de 0.994**, lo cual refleja un **excelente ajuste**.  
Todos los coeficientes principales son **estadísticamente significativos al 1 %**.  
El signo positivo de la autonomía y el signo negativo de su término cuadrático confirman la existencia de una **relación no lineal en forma de “∩”**: los incrementos iniciales en la autonomía fiscal elevan el gasto social, pero a niveles altos este efecto se atenúa.  
La proporción de transferencias también ejerce un **efecto positivo y significativo** sobre el gasto social, lo que sugiere que la dependencia financiera sigue siendo un factor clave para explicar el comportamiento del gasto municipal.

Los resultados detallados muestran que:
- El coeficiente positivo de **autonomía (229.06)** indica que, en niveles bajos, una mayor autonomía fiscal se asocia con un incremento del gasto social.  
- El coeficiente negativo de **autonomía² (-722.87)** refleja que este efecto se reduce progresivamente, evidenciando **rendimientos decrecientes**.  
- La variable **prop_transferencias (5.76)** mantiene un efecto positivo, sugiriendo que las transferencias intergubernamentales también contribuyen a aumentar el gasto social local.

El **R² ajustado (0.994)** muestra una mejora significativa respecto al modelo simple (0.961), confirmando que la relación no es estrictamente lineal y que el modelo complejo captura mejor las variaciones del gasto social.

**Predicción:**  
En comparación con el modelo simple, el **MSE promedio disminuyó**, lo que indica una mejora en la capacidad predictiva del modelo.  
Además, la desviación estándar de los errores de validación cruzada fue baja, mostrando que el rendimiento del modelo es **estable y consistente** entre los diferentes subconjuntos de datos.

En conjunto, la incorporación del término cuadrático mejora sustancialmente la capacidad explicativa y predictiva del modelo, representando de forma más realista cómo la autonomía financiera influye sobre el gasto social municipal.


```python
# =========================================================
# Gráfico del efecto cuadrático de la autonomía
# =========================================================

# Creamos un rango de valores de autonomía dentro de los observados
autonomia_range = np.linspace(df["autonomia"].min(), df["autonomia"].max(), 100)

# Calculamos la predicción manteniendo constante la proporción de transferencias (en su media)
prop_media = df["prop_transferencias"].mean()

# Generamos la variable cuadrática
autonomia2_range = autonomia_range**2

# Calculamos la predicción del ln(gasto_social) según el modelo complejo
pred_ln_gasto = (
    model_complex.params["const"]
    + model_complex.params["autonomia"] * autonomia_range
    + model_complex.params["autonomia2"] * autonomia2_range
    + model_complex.params["prop_transferencias"] * prop_media
)

# Convertimos de logaritmo a gasto social estimado
pred_gasto = np.expm1(pred_ln_gasto)

# --- Gráfico ---
plt.figure(figsize=(8,6))
plt.plot(autonomia_range, pred_gasto, color='blue', linewidth=2)
plt.title("Relación Predicha entre Autonomía Fiscal y Gasto Social")
plt.xlabel("Autonomía Fiscal")
plt.ylabel("Gasto Social Predicho")
plt.grid(True)
plt.show()
```

* La curva azul muestra la relación estimada por el modelo cuadrático, en la que
se observa un comportamiento claramente no lineal y con forma de campana.
Inicialmente, el gasto social predicho aumenta de forma acelerada conforme los municipios ganan mayor autonomía fiscal, alcanzando un punto máximo alrededor de una autonomía media (≈0.15).
A partir de ese punto, el gasto social proyectado comienza a disminuir, lo que sugiere que existe un nivel “óptimo” de autonomía fiscal a partir del cual los incrementos adicionales dejan de traducirse en más gasto social.
Esta evidencia puede interpretarse como una relación de eficiencia decreciente, donde los gobiernos locales más autónomos enfrentan limitaciones en su capacidad administrativa o en la priorización del gasto social.
En conjunto, el gráfico refleja la utilidad del modelo cuadrático para describir comportamientos económicos no lineales coherentes con la teoría fiscal.

```python
# --- Gráfico 1: Relación no lineal entre autonomía y gasto social ---
# Ordenar los valores por autonomía fiscal
df_sorted = df.sort_values("autonomia")

plt.figure(figsize=(8,6))
sns.scatterplot(x=df_sorted["autonomia"], y=df_sorted["ln_gasto_social"], alpha=0.6)

# --- Ajuste lineal ---
sns.lineplot(
    x=df_sorted["autonomia"],
    y=model_simple.predict(sm.add_constant(df_sorted[["autonomia", "prop_transferencias"]])),
    color='red', label='Ajuste lineal (modelo simple)'
)

# --- Ajuste cuadrático ---
sns.lineplot(
    x=df_sorted["autonomia"],
    y=model_complex.predict(sm.add_constant(df_sorted[["autonomia", "prop_transferencias", "autonomia2"]])),
    color='green', label='Ajuste cuadrático (modelo complejo)'
)

plt.title("Relación entre Autonomía y Gasto Social (lineal vs cuadrático)")
plt.xlabel("Autonomía fiscal")
plt.ylabel("ln(Gasto social)")
plt.legend()
plt.show()


```

* El gráfico permite observar la relación entre la autonomía fiscal y el gasto social en los gobiernos locales.
A partir de los datos, se aprecia una relación positiva: a medida que la autonomía fiscal aumenta, también lo hace el gasto social. Sin embargo, la curva del modelo cuadrático (verde) revela que esta relación no es estrictamente lineal, sino que presenta un comportamiento de rendimientos decrecientes.
En niveles bajos de autonomía, un incremento en la capacidad fiscal de los municipios se asocia con aumentos significativos del gasto social; no obstante, conforme la autonomía crece, el efecto marginal positivo se reduce, reflejando una posible saturación en la eficiencia del gasto o limitaciones institucionales.
Por su parte, el modelo lineal (rojo) subestima los cambios iniciales y sobreestima los finales, lo que muestra la conveniencia de incluir el término cuadrático para capturar mejor la realidad económica subyacente.

```python
# --- Gráfico 4: Residuos vs Ajustados (diagnóstico del modelo complejo) ---
fitted = model_complex.fittedvalues
resid = model_complex.resid

plt.figure(figsize=(7,5))
plt.scatter(fitted, resid, alpha=0.7, color="orchid")
plt.axhline(0, linestyle="--", color="gray")
plt.title("Residuos vs Valores ajustados (OLS complejo)")
plt.xlabel("Valores ajustados")
plt.ylabel("Residuos")
plt.grid(alpha=0.3)
plt.show()
```

* El gráfico de residuos permite analizar el comportamiento de los errores del modelo OLS complejo.
En general, los residuos se distribuyen alrededor de la línea cero, lo que indica que el modelo no presenta un sesgo sistemático en sus predicciones.
No obstante, se aprecia una ligera concentración de puntos en los valores altos del gasto social, lo que podría sugerir cierta heterocedasticidad leve (mayor variabilidad en los extremos).
Este patrón es esperable en datos fiscales y sociales, donde los municipios con mayores niveles de autonomía o gasto suelen presentar comportamientos más heterogéneos.
Aun así, la ausencia de un patrón claro (como una forma de U o una tendencia definida) confirma que la especificación del modelo es razonablemente adecuada y que los residuos se comportan de manera aleatoria en torno al valor esperado.

# ============================================
# Diagnóstico de Residuos
# ============================================

```python

from statsmodels.stats.diagnostic import het_breuschpagan, het_white
from statsmodels.stats.stattools import durbin_watson

# 1. Obtenemos los residuos del modelo complejo
residuos = model_complex.resid
predicciones = model_complex.fittedvalues

# --- Test de Breusch-Pagan ---
bp_test = het_breuschpagan(residuos, model_complex.model.exog)
labels = ["LM Statistic", "LM-Test p-value", "F-Statistic", "F-Test p-value"]
print("\n=== Test de Breusch-Pagan ===")
print(dict(zip(labels, bp_test)))

# --- Test de White ---
white_test = het_white(residuos, model_complex.model.exog)
print("\n=== Test de White ===")
print(dict(zip(labels, white_test)))

# --- Durbin-Watson (autocorrelación de residuos) ---
dw = durbin_watson(residuos)
print(f"\nDurbin-Watson Statistic: {dw:.4f}")

# --- Corrección de errores robustos (Newey-West HAC) ---
model_hac = model_complex.get_robustcov_results(cov_type='HAC', maxlags=1)
print("\n=== Resultados con Errores Robustos (Newey-West) ===")
print(model_hac.summary())
```

### Interpretación del Diagnóstico de Residuos

**Test de Breusch–Pagan:**  
El estadístico LM = 762.30 con p-valor ≈ 0.000 indica que se **rechaza la hipótesis nula de homocedasticidad**.  
Esto significa que los residuos presentan **heterocedasticidad**, es decir, la varianza de los errores no es constante entre las observaciones.  
Este resultado es habitual en datos de gasto público, donde los municipios más grandes o con mayor presupuesto tienden a mostrar variaciones más amplias.

**Test de White:**  
Los resultados confirman el hallazgo anterior (p-valor = 0.000), lo que refuerza la evidencia de **heterocedasticidad** en el modelo.

**Durbin–Watson (≈ 2.00):**  
El valor se encuentra muy cerca de 2, lo que indica **ausencia de autocorrelación serial** en los residuos.  
Esto cumple con uno de los supuestos fundamentales de los modelos de regresión OLS.

**Errores robustos (Newey–West):**  
Al estimar el modelo con errores robustos HAC (que corrigen heterocedasticidad y autocorrelación leve), los **coeficientes mantienen su significancia estadística y sus signos originales**.  
Esto demuestra que los resultados del modelo son **robustos**, incluso al ajustar los errores estándar.

En resumen, aunque existe heterocedasticidad —común en este tipo de datos—, los coeficientes siguen siendo estadísticamente significativos y el modelo mantiene su validez inferencial.  
El uso de errores robustos **corrige este problema** y confirma la **solidez del modelo OLS Complejo**.

**Multicolinealidad:**
El número de condición elevado (≈ 1.89e+03) sugiere la presencia de **multicolinealidad**, principalmente entre las variables `autonomía` y `autonomía²`, que están naturalmente correlacionadas.  
Sin embargo, esta multicolinealidad **no representa un problema grave** en este caso, ya que ambas variables son significativas y el modelo tiene un propósito explicativo y predictivo.  
En modelos polinómicos, este tipo de correlación es **esperada y aceptada**, pues permite capturar la relación no lineal sin afectar la validez de las conclusiones.


```python
# =========================================================
# 1 Validación temporal (TimeSeriesSplit)
# =========================================================

# Columna 'anio'
df_sorted = df.sort_values("anio").reset_index(drop=True)

# Variables explicativas y dependiente
X_time = df_sorted[["autonomia", "prop_transferencias", "autonomia2"]]
y_time = df_sorted["ln_gasto_social"]

# Configuración división temporal
tscv = TimeSeriesSplit(n_splits=5)

# MSE temporal (negativo → se multiplica por -1)
mse_time = -cross_val_score(
    LinearRegression(),
    X_time,
    y_time,
    cv=tscv,
    scoring="neg_mean_squared_error"
)

# R² temporal
r2_tscv = []
for train_index, test_index in tscv.split(X_time):
    X_train_t, X_test_t = X_time.iloc[train_index], X_time.iloc[test_index]
    y_train_t, y_test_t = y_time.iloc[train_index], y_time.iloc[test_index]
    model_t = LinearRegression().fit(X_train_t, y_train_t)
    y_pred_t = model_t.predict(X_test_t)
    r2_tscv.append(r2_score(y_test_t, y_pred_t))

mse_tscv = mse_time

print("\n--- Validación Cruzada Temporal (TimeSeriesSplit) ---")
print(f"MSE por fold (temporal): {mse_tscv}")
print(f"MSE promedio (temporal): {np.mean(mse_tscv):.4f} (+/- {np.std(mse_tscv):.4f})")
print(f"R² promedio (temporal): {np.mean(r2_tscv):.4f}")

# =========================================================
# Evaluación modelos Simple y Complejo (train/test)
# =========================================================

# Modelo OLS Simple
modelo_simple = LinearRegression()
modelo_simple.fit(X_train[["autonomia"]], y_train)
y_pred_simple_test = modelo_simple.predict(X_test[["autonomia"]])
mse_simple_test = mean_squared_error(y_test, y_pred_simple_test)
r2_simple_test = r2_score(y_test, y_pred_simple_test)

# =========================================================
# Tabla resumen de métricas
# =========================================================

resumen = pd.DataFrame({
    "Modelo": ["OLS Simple", "OLS Complejo", "OLS Complejo (TimeSeriesSplit)"],
    "MSE Promedio (CV)": [np.mean(mse_cv_simple), np.mean(mse_cv_complex), np.mean(mse_tscv)],
    "Desv. Std (CV)": [np.std(mse_cv_simple), np.std(mse_cv_complex), np.std(mse_tscv)],
    "MSE Test": [mse_simple_test, final_mse, np.mean(mse_tscv)],
    "R² Test": [r2_simple_test, final_r2, np.mean(r2_tscv)]
})

print("\n=== Resumen de Desempeño de los Modelos ===")
display(resumen.round(4))

# =========================================================
# Gráfico comparativo de desempeño
# =========================================================

plt.figure(figsize=(8,5))
sns.barplot(
    data=resumen,
    x="Modelo",
    y="MSE Promedio (CV)",
    palette="crest",
    alpha=0.9
)
plt.title("Comparación del MSE promedio (CV y TimeSeriesSplit)")
plt.ylabel("MSE promedio (menor es mejor)")
plt.xlabel("Modelo")
plt.grid(alpha=0.3, linestyle="--")
plt.show()

# =========================================================
# Interpretación final
# =========================================================

print("\n Interpretación:")
print("- El modelo OLS Complejo logra un MSE menor y un R² más alto tanto en validación cruzada como en test.")
print("- El modelo simple explica menos variación del gasto social (R² ≈ 0.96) frente al complejo (R² ≈ 0.99).")
print("- Con TimeSeriesSplit, se confirma que el modelo complejo mantiene su desempeño de forma estable en el tiempo.")
print("- Esto sugiere que la autonomía fiscal tiene un efecto cuadrático sobre el gasto social: a bajos niveles impulsa el gasto, pero luego se estabiliza o reduce su efecto marginal.")
```
